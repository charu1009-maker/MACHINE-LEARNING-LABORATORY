```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from google.colab import files

# Step 1: Upload training CSV
print("Upload Training CSV")
uploaded = files.upload()

train_file = list(uploaded.keys())[0]

# Step 2: Read CSV
data = pd.read_csv(train_file)

# Remove empty column
data = data.drop(columns=["Unnamed: 4"], errors="ignore")

print("\nDataset:")
print(data)

# Step 3: Select input and output
X = data[["Pothole", "Number of Potholes"]]
y = data["Level"]

# Step 4: Create Decision Tree
model = DecisionTreeClassifier(random_state=1)

# Step 5: Train model
model.fit(X, y)

print("\nModel trained successfully!")

# Step 6: Upload new CSV
print("\nUpload New Pothole CSV")
uploaded = files.upload()

test_file = list(uploaded.keys())[0]

# Step 7: Read new CSV
new_data = pd.read_csv(test_file)

# Step 8: Predict
new_data["Predicted Level"] = model.predict(
    new_data[["Pothole", "Number of Potholes"]]
)

# Step 9: Display result
print("\nPothole Severity Prediction:")
print(new_data)

# Step 10: Save output
new_data.to_csv("pothole_prediction.csv", index=False)

print("\nOutput saved as pothole_prediction.csv")

# Step 11: Download output
files.download("pothole_prediction.csv")
```
